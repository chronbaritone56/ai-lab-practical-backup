## Practical 2: Breadth-First Search and Depth-First Search

**Aim / explanation:** BFS explores nearby nodes first using a queue. DFS follows one path as far as possible before backtracking, using a stack or recursion.

**Before running:** Read the code and try to predict what it will print.

```python
# Practical 2: BFS and DFS on a graph

from collections import deque

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}

def bfs(graph, start):
    visited = []
    queue = deque([start])
    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.append(node)
            queue.extend(neighbour for neighbour in graph[node]
                         if neighbour not in visited)
    return visited

def dfs(graph, start):
    visited = []
    stack = [start]
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.append(node)
            # Reverse so the leftmost neighbour is visited first
            stack.extend(reversed(graph[node]))
    return visited

print("BFS:", bfs(graph, "A"))
print("DFS:", dfs(graph, "A"))
```

**Expected output / note**

BFS: ['A', 'B', 'C', 'D', 'E', 'F']
DFS: ['A', 'B', 'D', 'E', 'C', 'F']

**Viva questions**
- **Q:** What is BFS? **A:** Breadth-First Search visits nodes level by level.
- **Q:** Which data structure does BFS use? **A:** A queue.
- **Q:** Which data structure can DFS use? **A:** A stack or recursion.


## Practical 3: A* Search Algorithm

**Aim / explanation:** A* finds a low-cost path by combining the cost already travelled (g) with an estimated cost to the goal (h). The estimate is called a heuristic.

**Before running:** Read the code and try to predict what it will print.

```python
# Practical 3: A* search on a small weighted graph
import heapq

graph = {
    "A": {"B": 1, "C": 4},
    "B": {"C": 2, "D": 5},
    "C": {"D": 1},
    "D": {}
}
# Estimated remaining distance to goal D
heuristic = {"A": 4, "B": 3, "C": 1, "D": 0}

def astar(start, goal):
    frontier = [(heuristic[start], 0, start, [start])]
    best_cost = {start: 0}

    while frontier:
        f_score, cost, node, path = heapq.heappop(frontier)
        if node == goal:
            return path, cost

        if cost > best_cost.get(node, float("inf")):
            continue

        for neighbour, step_cost in graph[node].items():
            new_cost = cost + step_cost
            if new_cost < best_cost.get(neighbour, float("inf")):
                best_cost[neighbour] = new_cost
                priority = new_cost + heuristic[neighbour]
                heapq.heappush(frontier,
                               (priority, new_cost, neighbour,
                                path + [neighbour]))
    return None, float("inf")

path, cost = astar("A", "D")
print("Best path:", path)
print("Total cost:", cost)
```

**Expected output / note**

Best path: ['A', 'B', 'C', 'D']
Total cost: 4

**Viva questions**
- **Q:** What does A* do? **A:** It searches for a low-cost path from a start node to a goal.
- **Q:** What is g(n)? **A:** The cost from the start to the current node.
- **Q:** What is h(n)? **A:** An estimate of the remaining cost to the goal.


## Practical 4: Tic-Tac-Toe using Minimax

**Aim / explanation:** Minimax is a decision-making algorithm for two-player games. One player tries to maximise the score while the opponent tries to minimise it.

**Before running:** Read the code and try to predict what it will print.

```python
# Practical 4: A small Minimax example for Tic-Tac-Toe
# Board cells are numbered 0 to 8.
WIN_LINES = [(0,1,2), (3,4,5), (6,7,8),
             (0,3,6), (1,4,7), (2,5,8),
             (0,4,8), (2,4,6)]

def winner(board):
    for a, b, c in WIN_LINES:
        if board[a] == board[b] == board[c] and board[a] != " ":
            return board[a]
    return None

def minimax(board, is_maximizing):
    result = winner(board)
    if result == "X":
        return 1
    if result == "O":
        return -1
    if " " not in board:
        return 0

    if is_maximizing:
        best = -float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                best = max(best, minimax(board, False))
                board[i] = " "
        return best
    else:
        best = float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                best = min(best, minimax(board, True))
                board[i] = " "
        return best

def best_move(board):
    best_score = -float("inf")
    move = None
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            score = minimax(board, False)
            board[i] = " "
            if score > best_score:
                best_score, move = score, i
    return move

board = ["X", "O", "X",
         "O", "X", " ",
         " ", " ", "O"]
move = best_move(board)
print("Best position for X (0-8):", move)
board[move] = "X"
print("Board after move:")
for row in range(0, 9, 3):
    print(board[row:row+3])
```

**Expected output / note**

Best position for X (0-8): 6
Board after move:
['X', 'O', 'X']
['O', 'X', ' ']
['X', ' ', 'O']

**Viva questions**
- **Q:** What is Minimax? **A:** A search method for choosing moves in two-player games.
- **Q:** What does the maximising player do? **A:** Chooses the move with the highest score.
- **Q:** Why is the board restored after testing a move? **A:** So each possible move is evaluated from the same original position.


## Practical 5: Knowledge Representation using Facts and Rules

**Aim / explanation:** Knowledge representation stores information in a form a computer can use. This example uses simple facts and rules to infer a new fact.

**Before running:** Read the code and try to predict what it will print.

```python
# Practical 5: Knowledge representation using simple rules

facts = {
    ("is_a", "sparrow", "bird"),
    ("has", "sparrow", "wings"),
    ("is_a", "penguin", "bird")
}

def infer_can_fly(animal):
    # Simple rule: birds can fly, except penguins
    is_bird = ("is_a", animal, "bird") in facts
    is_penguin = animal == "penguin"
    return is_bird and not is_penguin

for animal in ["sparrow", "penguin"]:
    print(animal, "can fly:", infer_can_fly(animal))
```

**Expected output / note**

sparrow can fly: True
penguin can fly: False

**Viva questions**
- **Q:** What is knowledge representation? **A:** A way of storing facts and relationships so a computer can use them.
- **Q:** What is a rule? **A:** A condition that lets a system draw a conclusion.
- **Q:** Is this example a complete real-world bird model? **A:** No. It is a small demonstration of facts and rules.


## Practical 6: Bayes' Theorem / Simple Naive Bayes

**Aim / explanation:** Bayes' theorem updates the probability of a hypothesis after seeing evidence. This simple example calculates the probability of rain given that the ground is wet, using made-up probabilities.

**Before running:** Read the code and try to predict what it will print.

```python
# Practical 6: Bayes' theorem with example probabilities

# P(Rain) = 0.30
# P(Wet | Rain) = 0.80
# P(Wet | No Rain) = 0.20
p_rain = 0.30
p_wet_given_rain = 0.80
p_wet_given_no_rain = 0.20

p_wet = (p_wet_given_rain * p_rain
         + p_wet_given_no_rain * (1 - p_rain))

p_rain_given_wet = (p_wet_given_rain * p_rain) / p_wet

print("P(Wet):", round(p_wet, 2))
print("Probability of rain given wet ground:",
      round(p_rain_given_wet, 2))
```

**Expected output / note**

P(Wet): 0.38
Probability of rain given wet ground: 0.63

**Viva questions**
- **Q:** What is Bayes' theorem used for? **A:** To update a probability using new evidence.
- **Q:** What does P(A|B) mean? **A:** The probability of A given that B has happened.
- **Q:** What is Naive Bayes? **A:** A probabilistic classifier that assumes features are conditionally independent given the class.


## Practical 7: Linear Regression and Decision Tree

**Aim / explanation:** Linear regression predicts a numeric value. A decision tree learns rules that can be used for classification or prediction. This example uses scikit-learn.

## Practical 8: K-Means Clustering

**Aim / explanation:** K-Means is an unsupervised learning algorithm. It groups data points into K clusters based on their similarity.

**Before running:** Read the code and try to predict what it will print.

```python
# Practical 8: K-Means clustering
from sklearn.cluster import KMeans

points = [[1, 2], [1, 4], [1, 0],
          [10, 2], [10, 4], [10, 0]]

model = KMeans(n_clusters=2, random_state=0, n_init=10)
labels = model.fit_predict(points)

print("Cluster labels:", labels.tolist())
print("Cluster centres:")
for centre in model.cluster_centers_:
    print([round(value, 2) for value in centre])
```

**Expected output / note**

Cluster labels may be [1, 1, 1, 0, 0, 0] or the same groups with labels swapped.
Cluster centres are approximately [1, 2] and [10, 2].

**Viva questions**
- **Q:** What is clustering? **A:** Grouping similar data points together.
- **Q:** Is K-Means supervised or unsupervised? **A:** Unsupervised.
- **Q:** What does K represent? **A:** The number of clusters to create.


## Practical 9: Basic Natural Language Processing

**Aim / explanation:** Natural Language Processing (NLP) helps computers process human language. This beginner example converts text to lowercase, removes punctuation, and counts words.

**Before running:** Read the code and try to predict what it will print.

```python
# Practical 9: Basic NLP without extra downloads
import re
from collections import Counter

text = "AI is useful. AI helps people, and AI can learn!"

clean_text = re.sub(r"[^a-zA-Z\s]", "", text.lower())
words = clean_text.split()
counts = Counter(words)

print("Cleaned words:", words)
print("Word counts:", dict(counts))
print("Most common word:", counts.most_common(1)[0][0])
```

**Expected output / note**

Cleaned words: ['ai', 'is', 'useful', 'ai', 'helps', 'people', 'and', 'ai', 'can', 'learn']
Word counts: {'ai': 3, 'is': 1, 'useful': 1, 'helps': 1, 'people': 1, 'and': 1, 'can': 1, 'learn': 1}
Most common word: ai

**Viva questions**
- **Q:** What is NLP? **A:** A field of AI that enables computers to work with human language.
- **Q:** What is tokenisation? **A:** Splitting text into smaller units such as words.
- **Q:** Why convert text to lowercase? **A:** To treat words like 'AI' and 'ai' as the same token.
